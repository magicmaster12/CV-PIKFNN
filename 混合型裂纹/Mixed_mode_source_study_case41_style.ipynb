# KM–PIKFNN 混合型中心斜裂纹：同时求解 $K_I$ 与 $K_{II}$

实现用户所给文献 §5.4 / Fig.15 / Table 5 的算例：板宽 $H=4$、板高 $2H=8$、中心裂纹半长 $a=1$，平面应力，$\nu=0.3$，上下边均匀拉应力取 $\sigma_0=1$，左右边无牵引。裂纹与水平轴夹角为 $15^\circ,30^\circ,45^\circ,60^\circ$。所有数值采用一致单位，$K$ 的量纲为应力乘长度平方根。

延续一型、二型程序：外部基本解 + 首阶及高阶 KM–Williams 富集，列归一化，SVD 热启动，可选 Adam/L-BFGS，按最低损失恢复系数。**不生成裂尖源点圈，不把参考 SIF 放入损失、初始化或参数选择。**

斜裂纹不适用一/二型的四分之一模型。这里沿裂纹延长线分为 $\Omega_1$ 与 $\Omega_2$，利用中心反演对称性求解上方子域；两侧韧带严格匹配位移和牵引。子域的人工源点边界位于该子域外部；其中底部源点可位于另一物理子域内，但不参与该子域的场评价。

```python
import math
import time
from pathlib import Path
import numpy as np
import matplotlib.pyplot as plt
from scipy.linalg import lstsq
try:
    import torch
except ImportError:
    torch = None

cfg = {
    "a":1.0, "H":4.0, "E":1.0, "nu":0.3, "sigma0":1.0,
    "angles_deg":(15,30,45,60),
    "n_source_per_edge":80, "source_offset":0.3,
    "adaptive_sources":True, "source_offset_bounds":(0.02,1.20),
    "source_initial_values":((0.10,0.10),(0.30,0.30),(0.60,0.30),(0.90,0.60)),
    "source_lr":1e-3, "source_log_every":10,

    "n_outer":151, "n_crack":151, "n_interface_each":121,
    "higher_orders":(1.5,2.5),
    "lambda_reg":1e-10, "svd_rcond":1e-12,
    "ls_warm_start":False, "adam_steps":0, "adam_lr":1e-3,
    "lbfgs_epochs":200, "lbfgs_max_iter":20,
    "plot_angle":30, "nx_plot":101, "ny_plot":201,
    "save_outputs":False, "output_dir":"mixed_mode_adaptive_cold_results", "seed":2026,
}
a,H,E,nu,sigma0=(float(cfg[k]) for k in ("a","H","E","nu","sigma0"))
mu=E/(2*(1+nu));kappa=(3-nu)/(1+nu)  # Plane stress
lam=E*nu/(1-nu**2)                    # Reduced in-plane Lame constant
assert 0<a<H/2 and E>0 and -1<nu<0.5
np.random.seed(cfg["seed"])
if torch is not None:
    torch.set_default_dtype(torch.float64)
    torch.manual_seed(cfg["seed"])
    device=torch.device("cuda" if torch.cuda.is_available() else "cpu")
else:
    device="cpu"
print("Complex basis: CPU NumPy; coefficient training:",
      device if torch is not None else "SciPy SVD only (PyTorch unavailable)")

# Right-tip references transcribed from the supplied Table 5.
# These dictionaries are read ONLY after the boundary problem is solved.
BEM_REF={15:(1.9705,0.4641),30:(1.6020,0.8180),45:(1.0838,0.9666),60:(0.5494,0.8551)}
PINN_REF={15:(1.9715,0.4696),30:(1.6058,0.8205),45:(1.0904,0.9618),60:(0.5441,0.8592)}

def maybe_save(fig,name):
    if cfg["save_outputs"]:
        folder=Path(cfg["output_dir"]);folder.mkdir(parents=True,exist_ok=True)
        fig.savefig(folder/name,dpi=600,bbox_inches="tight")
if torch is None:
    raise ImportError("Adaptive source training requires PyTorch")
# Complex source autograd on CPU avoids CUDA/NVRTC compatibility issues.
device=torch.device("cpu")
torch.set_num_threads(min(4,torch.get_num_threads()))
print("Adaptive source geometry and training: CPU float64")
import copy
import csv
```

## 复势、坐标旋转与裂尖系数
每个裂尖采用向裂纹前方指向的局部坐标 $w=e^{-i\alpha}(z-z_t)$。右尖 $\alpha=\theta$，左尖 $\alpha=\theta+\pi$。
\[
\phi_p^I=\frac{H^{1/2-p}}{\sqrt{2\pi}}w^p,\quad\psi_p^I=(1-p)\phi_p^I,
\qquad
\phi_p^{II}=-\frac{iH^{1/2-p}}{\sqrt{2\pi}}w^p,\quad\psi_p^{II}=-(p+1)\phi_p^{II}.
\]
在每个裂尖局部原点直接应用 KM 公式，再将位移向量、应力张量旋转回整体坐标，因此不需要遗漏或另补整体坐标的平移项。$p=1/2$ 单位场的系数分别为 $K_I,K_{II}$。最后四列顺序固定为 **右尖 KI、右尖 KII、左尖 KI、左尖 KII**。

对 $\Omega_1$，右尖取 $0\leq\arg w\leq\pi$，左尖取 $-\pi\leq\arg w\leq0$。代码显式指定裂面分支，避免浮点零的符号导致错误裂面值。

```python
def km(z,phi,dp,ddp,psi,dpsi):
 s=4*dp.real;d=2*(z.conj()*ddp+dpsi);u=(kappa*phi-z*dp.conj()-psi.conj())/(2*mu)
 return np.array([u.real,u.imag,(s-d.real)/2,(s+d.real)/2,d.imag/2]).T

def force(z,s,P):
 q=z-s;den=2*np.pi*(kappa+1)
 phi=-P/den*np.log(q)+P/(2*np.pi*kappa*(kappa+1));dp=-P/den/q;ddp=P/den/q**2
 psi=kappa*P.conjugate()/den*np.log(q)+P*s.conjugate()/den/q
 dpsi=kappa*P.conjugate()/den/q-P*s.conjugate()/den/q**2
 return km(z,phi,dp,ddp,psi,dpsi)

def rotate(f,angle):
 c,s=np.cos(angle),np.sin(angle);ux,uy,xx,yy,xy=f.T
 return np.array([c*ux-s*uy,s*ux+c*uy,c*c*xx+s*s*yy-2*c*s*xy,s*s*xx+c*c*yy+2*c*s*xy,c*s*(xx-yy)+(c*c-s*s)*xy]).T

def will(z,angle,side,mode,p):
 # right angle, left angle+pi; local w origin at each tip
 tip=side*a*np.exp(1j*angle);orientation=angle+(np.pi if side==-1 else 0)
 w=(z-tip)*np.exp(-1j*orientation)
 # enforce intended upper/lower face branch despite floating-point rotation noise
 phase = np.arctan2(np.maximum(w.imag,0),w.real) if side==1 else -np.arctan2(np.maximum(-w.imag,0),w.real)
 def power(q):return np.abs(w)**q*np.exp(1j*q*phase)
 C=(1 if mode=='I' else -1j)/np.sqrt(2*np.pi)*H**(.5-p)
 phi=C*power(p);dp=C*p*power(p-1);ddp=C*p*(p-1)*power(p-2)
 factor=(1-p) if mode=='I' else -(p+1)
 return rotate(km(w,phi,dp,ddp,factor*phi,factor*dp),orientation)

def solve(theta=30,n=80,nb=151,nface=151,nint=121,orders=(1.5,2.5),margin=.3,reg=1e-10,verbose=True):
 ang=np.deg2rad(theta);c,s=np.cos(ang),np.sin(ang);lim=H/2/c
 vertices=np.array([-H/2-1j*H/2*np.tan(ang),H/2+1j*H/2*np.tan(ang),H/2+1j*H,-H/2+1j*H])
 # convex polygon offset to create continuous exterior artificial boundary
 edges=np.roll(vertices,-1)-vertices; normals=-1j*edges/abs(edges)
 vo=[]
 for i in range(4):
  # intersect two offset lines at vertex
  e1=edges[i-1];e2=edges[i];p1=vertices[i]+margin*normals[i-1];p2=vertices[i]+margin*normals[i]
  M=np.array([[e1.real,-e2.real],[e1.imag,-e2.imag]])
  t=np.linalg.solve(M,np.array([(p2-p1).real,(p2-p1).imag]))[0]
  vo.append(p1+t*e1)
 vo=np.array(vo);src=np.concatenate([vo[i]+(np.arange(n)+.5)/n*(vo[(i+1)%4]-vo[i]) for i in range(4)])
 def line(v1,v2,N):return v1+(np.arange(N)+.5)/N*(v2-v1)
 pts={'top':line(vertices[3],vertices[2],nb),'right':line(vertices[1],vertices[2],nb),'left':line(vertices[0],vertices[3],nb),'crack':line(-.999*a,.999*a,nface)*np.exp(1j*ang),'int':np.r_[line(-lim,-1.001*a,nint),line(1.001*a,lim,nint)]*np.exp(1j*ang)}
 def basis(z):
  cols=[force(z,ss,P) for ss in src for P in (1+0j,1j)]
  cols +=[will(z,ang,side,mode,p) for p in orders for side in (1,-1) for mode in ('I','II')]
  cols +=[will(z,ang,side,mode,.5) for side in (1,-1) for mode in ('I','II')]
  return np.stack(cols,axis=2)
 rows=[];rhs=[]
 def add(block,target=0,weight=1):
  f=np.sqrt(weight/len(block));rows.append(f*block);rhs.append(np.full(len(block),f*target))
 for key in ['top','right','left']:
  mat=basis(pts[key]);add(mat[:,3 if key=='top' else 2],sigma0 if key=='top' else 0);add(mat[:,4])
 mat=basis(pts['crack']);nx,ny=-s,c
 tx=nx*mat[:,2]+ny*mat[:,4];ty=nx*mat[:,4]+ny*mat[:,3]
 add(tx,weight=4);add(ty,weight=4)
 z=pts['int'];m=basis(z);mm=basis(-z)
 add(m[:,0]+mm[:,0],weight=2);add(m[:,1]+mm[:,1],weight=2)
 add(nx*(m[:,2]-mm[:,2])+ny*(m[:,4]-mm[:,4]),weight=2)
 add(nx*(m[:,4]-mm[:,4])+ny*(m[:,3]-mm[:,3]),weight=2)
 # Pure traction problem retains a rigid rotation; fix one gauge only.
 add(basis(np.array([1j*H]))[:,0],weight=1)
 A=np.vstack(rows);y=np.concatenate(rhs);scale=np.linalg.norm(A,axis=0);An=A/scale
 R=np.eye(An.shape[1])[:-4]*np.sqrt(reg/(An.shape[1]-4))
 tt=np.zeros(An.shape[1]);rank=0;coeff=tt/scale  # Cold initialization; no SVD solve
 loss=np.linalg.norm(A@coeff-y)**2
 if verbose:print(theta,'n',n,'K right/left',coeff[-4:],'loss',loss,flush=True)
 return dict(theta=theta,coeff=coeff,src=src,vertices=vertices,source_vertices=vo,basis=basis,points=pts,A=An,rhs=y,scale=scale,loss=loss,theta_svd=tt,rank=int(rank),reg=reg,orders=orders,angle_rad=ang,margin=margin,nb=nb,nface=nface,nint=nint)
```

## 连接条件与唯一性
中心反演对称性给出 $\mathbf u_2(z)=-\mathbf u_1(-z)$，$\boldsymbol\sigma_2(z)=\boldsymbol\sigma_1(-z)$。裂纹延长线、裂纹外的两段韧带上施加
\[
\mathbf u_1(z)+\mathbf u_1(-z)=0,\qquad
[\boldsymbol\sigma_1(z)-\boldsymbol\sigma_1(-z)]\mathbf n=0.
\]
裂面施加 $\boldsymbol\sigma_1\mathbf n=0$。所有外边界按整体坐标的法向施加拉伸或自由牵引。纯牵引问题保留一个刚体转动，以上边中点 $u_x=0$ 固定转动规范；该条件不影响应力和 SIF。

```python
# Differentiable source boundary: x/y edge offsets in physical length units.
def offset_polygon(vertices,margin_x,margin_y):
    edges=np.roll(vertices,-1)-vertices;normals=-1j*edges/abs(edges)
    offsets=np.array([margin_y,margin_x,margin_y,margin_x])
    result=[]
    for i in range(4):
        e1,e2=edges[i-1],edges[i]
        p1=vertices[i]+offsets[i-1]*normals[i-1]
        p2=vertices[i]+offsets[i]*normals[i]
        mat=np.array([[e1.real,-e2.real],[e1.imag,-e2.imag]])
        t=np.linalg.solve(mat,np.array([(p2-p1).real,(p2-p1).imag]))[0]
        result.append(p1+t*e1)
    return np.asarray(result)


def adaptive_force_matrix(z,sources):
    z=torch.as_tensor(z,dtype=torch.complex128)[:,None]
    q=z-sources[None,:];den=2*np.pi*(kappa+1)
    logq=torch.complex(torch.log(torch.abs(q)),torch.atan2(q.imag,q.real))
    results=[]
    for force_value in (1+0j,1j):
        P=torch.tensor(force_value,dtype=torch.complex128)
        phi=-P/den*logq+P/(2*np.pi*kappa*(kappa+1));dp=-P/(den*q);ddp=P/(den*q*q)
        psi=kappa*P.conj()/den*logq+P*sources.conj()[None,:]/(den*q)
        dpsi=kappa*P.conj()/(den*q)-P*sources.conj()[None,:]/(den*q*q)
        total=4*dp.real;diff=2*(z.conj()*ddp+dpsi)
        u=(kappa*phi-z*dp.conj()-psi.conj())/(2*mu)
        results.append(torch.stack((u.real,u.imag,(total-diff.real)/2,(total+diff.real)/2,diff.imag/2),1))
    return torch.stack(results,3).reshape(len(z),5,-1)


class AdaptiveMixedModel(torch.nn.Module):
    def __init__(self,case,initial):
        super().__init__();self.case=case
        lo,hi=cfg['source_offset_bounds']
        values=torch.tensor(initial,dtype=torch.float64)
        if torch.any(values<=lo) or torch.any(values>=hi):raise ValueError('Initial source offsets must lie inside bounds')
        self.raw=torch.nn.Parameter(torch.logit((values-lo)/(hi-lo)),requires_grad=cfg['adaptive_sources'])
        self.lo,self.hi=lo,hi
        vertices=case['vertices']
        self.register_buffer('vertices',torch.tensor(vertices,dtype=torch.complex128))
        self.register_buffer('dx',torch.tensor(offset_polygon(vertices,1.,0.)-vertices,dtype=torch.complex128))
        self.register_buffer('dy',torch.tensor(offset_polygon(vertices,0.,1.)-vertices,dtype=torch.complex128))
        self.n=len(case['src'])//4
        self.t=(torch.arange(self.n,dtype=torch.float64)+.5)/self.n
        pts=case['points'];self.keys=['top','right','left','crack','int','minus_int','gauge']
        groups=[pts[k] for k in self.keys[:5]]+[-pts['int'],np.array([1j*H])]
        self.lengths=[len(z) for z in groups];self.z=np.concatenate(groups)
        # Williams enrichment does not depend on source locations.
        fixed=case['basis'](self.z)[:,:,2*len(case['src']):]
        self.register_buffer('fixed',torch.tensor(fixed,dtype=torch.float64))
        # Freeze initial column normalization throughout geometry training.
        with torch.no_grad():A,y=self.system_unscaled()
        self.register_buffer('scale',torch.clamp(torch.linalg.norm(A,dim=0),min=1e-14))
        self.register_buffer('rhs',y)
        self.theta=torch.nn.Parameter(torch.zeros(A.shape[1],dtype=torch.float64))

    def offsets(self):return self.lo+(self.hi-self.lo)*torch.sigmoid(self.raw)
    def source_vertices(self):
        mx,my=self.offsets();return self.vertices+mx*self.dx+my*self.dy
    def sources(self):
        v=self.source_vertices();return (v[:,None]+self.t[None,:]*(torch.roll(v,-1)-v)[:,None]).reshape(-1)
    def system_unscaled(self):
        mat=torch.cat((adaptive_force_matrix(self.z,self.sources()),self.fixed),2)
        m=dict(zip(self.keys,torch.split(mat,self.lengths)))
        rows=[];targets=[]
        def add(block,target=0.,weight=1.):
            f=math.sqrt(weight/len(block));rows.append(f*block)
            targets.append(torch.full((len(block),),f*target))
        for key in ['top','right','left']:
            add(m[key][:,3 if key=='top' else 2],sigma0 if key=='top' else 0.)
            add(m[key][:,4])
        nx,ny=-np.sin(self.case['angle_rad']),np.cos(self.case['angle_rad'])
        v=m['crack'];add(nx*v[:,2]+ny*v[:,4],weight=4);add(nx*v[:,4]+ny*v[:,3],weight=4)
        v,w=m['int'],m['minus_int']
        add(v[:,0]+w[:,0],weight=2);add(v[:,1]+w[:,1],weight=2)
        add(nx*(v[:,2]-w[:,2])+ny*(v[:,4]-w[:,4]),weight=2)
        add(nx*(v[:,4]-w[:,4])+ny*(v[:,3]-w[:,3]),weight=2)
        add(m['gauge'][:,0])
        return torch.cat(rows),torch.cat(targets)
    def losses(self):
        A,_=self.system_unscaled();r=(A/self.scale)@self.theta-self.rhs
        boundary=torch.sum(r*r);reg=self.case['reg']*torch.mean(self.theta[:-4]**2)
        return boundary+reg,boundary,reg
    def coefficients(self):return self.theta/self.scale


def objective_numpy(case,theta):
    return float(np.sum((case['A']@np.asarray(theta)-case['rhs'])**2)+case['reg']*np.mean(np.asarray(theta)[:-4]**2))


def train_case(case,initial=None,verbose=True):
    initial=(case['margin'],case['margin']) if initial is None else initial
    net=AdaptiveMixedModel(case,initial)
    with torch.no_grad():A,_=net.system_unscaled();An=(A/net.scale).numpy()
    assert torch.count_nonzero(net.theta).item()==0  # includes all four SIF coefficients
    rows=[];best=float('inf');best_state=None;best_origin='initial';best_record=0
    # Reference values are reporting only and do not enter losses/initialization.
    refs=np.asarray(BEM_REF[case['theta']])*sigma0
    t0=time.perf_counter()
    def record(stage,iteration,display=False):
        nonlocal best,best_state,best_origin,best_record
        with torch.no_grad():
            total,boundary,reg=[float(v) for v in net.losses()]
            mx,my=net.offsets().tolist();ki,kii=net.coefficients()[-4:-2].tolist()
        row=dict(record=len(rows),stage=stage,iteration=iteration,offset_x=mx,offset_y=my,
                 offset_x_over_H=mx/H,offset_y_over_H=my/H,KI=ki,KII=kii,
                 err_KI=abs(ki-refs[0])/abs(refs[0]),err_KII=abs(kii-refs[1])/abs(refs[1]),
                 weighted_boundary_loss=boundary,boundary_residual=np.sqrt(boundary),total_loss=total,regularization=reg)
        rows.append(row)
        if total<best:
            best=total;best_state=copy.deepcopy(net.state_dict());best_origin=f'{stage} {iteration}';best_record=row['record']
        if display:print(f"{stage} {iteration}: mx={mx:.6f}, my={my:.6f}, KI={ki:.8f}, KII={kii:.8f}, "
                          f"boundary={boundary:.3e}, errKI={row['err_KI']:.3%}, errKII={row['err_KII']:.3%}",flush=True)
    record('zero initialization',0,verbose)
    multiplier=1/max(best,1e-12) # fixed loss scaling; source parameters and coefficients are optimized jointly
    if cfg['adam_steps']:
        params=[{'params':[net.theta],'lr':cfg['adam_lr']}]
        if cfg['adaptive_sources']:params.append({'params':[net.raw],'lr':cfg['source_lr']})
        opt=torch.optim.Adam(params)
        for epoch in range(1,cfg['adam_steps']+1):
            opt.zero_grad();loss,_,_=net.losses();(loss*multiplier).backward();opt.step()
            record('Adam',epoch,verbose and (epoch==1 or epoch%cfg['source_log_every']==0))
    net.load_state_dict(best_state)
    opt=torch.optim.LBFGS([p for p in net.parameters() if p.requires_grad],lr=1.,
        max_iter=cfg['lbfgs_max_iter'],max_eval=cfg['lbfgs_max_iter']+10,history_size=100,
        tolerance_grad=1e-12,tolerance_change=1e-14,line_search_fn='strong_wolfe')
    for epoch in range(1,cfg['lbfgs_epochs']+1):
        def closure():
            opt.zero_grad();loss,_,_=net.losses();value=loss*multiplier;value.backward();return value
        opt.step(closure);record('L-BFGS',epoch,verbose and (epoch==1 or epoch%cfg['source_log_every']==0 or epoch==cfg['lbfgs_epochs']))
    net.load_state_dict(best_state)
    with torch.no_grad():
        A,_=net.system_unscaled()
        case.update(src=net.sources().numpy().copy(),source_vertices=net.source_vertices().numpy().copy(),
                    scale=net.scale.numpy().copy(),A=(A/net.scale).numpy().copy(),rhs=net.rhs.numpy().copy(),
                    theta_final=net.theta.numpy().copy(),coeff=net.coefficients().numpy().copy(),
                    offset_x=float(net.offsets()[0]),offset_y=float(net.offsets()[1]))
    # Rebuild postprocessing closure with optimized sources; no SVD re-solve.
    src=case['src'].copy();ang=case['angle_rad'];orders=case['orders']
    def final_basis(z):
        z=np.asarray(z,dtype=np.complex128).reshape(-1)
        cols=[force(z,s,P) for s in src for P in (1+0j,1j)]
        cols +=[will(z,ang,side,mode,p) for p in orders for side in (1,-1) for mode in ('I','II')]
        cols +=[will(z,ang,side,mode,.5) for side in (1,-1) for mode in ('I','II')]
        return np.stack(cols,axis=2)
    case.update(basis=final_basis,best_origin=best_origin,best_record=best_record,
                history=[r['total_loss'] for r in rows],adaptive_history=rows,
                initial_offsets=tuple(initial),loss=rows[best_record]['weighted_boundary_loss'],
                elapsed=time.perf_counter()-t0)
    folder=Path(cfg['output_dir']);folder.mkdir(parents=True,exist_ok=True)
    stem=f"angle_{case['theta']}deg_mx_{initial[0]:g}_my_{initial[1]:g}"
    with (folder/f'{stem}_history.csv').open('w',newline='',encoding='utf-8-sig') as f:
        writer=csv.DictWriter(f,fieldnames=list(rows[0]));writer.writeheader();writer.writerows(rows)
    np.savez(folder/f'{stem}_best.npz',sources=src,source_vertices=case['source_vertices'],
             coefficients=case['coeff'],initial_offsets=initial,final_offsets=[case['offset_x'],case['offset_y']],
             KI=case['coeff'][-4],KII=case['coeff'][-3],best_record=best_record)
    return case


cases={}
for angle in cfg['angles_deg']:
    t0=time.perf_counter()
    case=solve(angle,n=cfg['n_source_per_edge'],nb=cfg['n_outer'],nface=cfg['n_crack'],
               nint=cfg['n_interface_each'],orders=cfg['higher_orders'],margin=cfg['source_offset'],
               reg=cfg['lambda_reg'],verbose=False)
    cases[angle]=train_case(case)
    case['elapsed']=time.perf_counter()-t0
    print(f"angle={angle}: adopted {case['best_origin']}, KI={case['coeff'][-4]:.8f}, KII={case['coeff'][-3]:.8f}")
```

## 域内本构核查、独立边界点与裂尖极限
使用新的边界配点核查残差。右裂尖前方的局部 $\sigma_{nn}$ 与 $\sigma_{tn}$ 分别提取 $K_I,K_{II}$；上裂面局部 $u_n,u_t$ 的 $\sqrt r$ 系数提供第二种独立提取。位移拟合包含常数项，防止刚体分量污染裂尖极限。

```python
def evaluate_upper(case,z,batch_size=1200):
    z=np.asarray(z,dtype=np.complex128).reshape(-1)
    parts=[]
    for start in range(0,len(z),batch_size):
        mat=case["basis"](z[start:start+batch_size])
        parts.append(np.einsum("nqc,c->nq",mat,case["coeff"]))
    return np.vstack(parts)

def evaluate_full(case,z,batch_size=1200):
    z=np.asarray(z,dtype=np.complex128).reshape(-1)
    local=z*np.exp(-1j*case["angle_rad"])
    upper=local.imag>=0
    out=np.empty((len(z),5))
    if upper.any():out[upper]=evaluate_upper(case,z[upper],batch_size)
    if (~upper).any():
        lower=evaluate_upper(case,-z[~upper],batch_size)
        lower[:,:2]*=-1
        out[~upper]=lower
    return out

def independent_boundary_checks(case,N=203):
    v=case["vertices"];ang=case["angle_rad"];nx,ny=-np.sin(ang),np.cos(ang)
    def line(p,q,n):return p+(np.arange(n)+.37)/n*(q-p)
    residuals={}
    for name,p,q in (("top",v[3],v[2]),("right",v[1],v[2]),("left",v[0],v[3])):
        f=evaluate_upper(case,line(p,q,N))
        target=sigma0 if name=="top" else 0
        vals=np.c_[f[:,3 if name=="top" else 2]-target,f[:,4]]
        residuals[name+"_traction_rms"]=float(np.sqrt(np.mean(vals**2)))
    z=line(-.999*a,.999*a,N)*np.exp(1j*ang)
    f=evaluate_upper(case,z)
    vals=np.c_[nx*f[:,2]+ny*f[:,4],nx*f[:,4]+ny*f[:,3]]
    residuals["crack_traction_rms"]=float(np.sqrt(np.mean(vals**2)))
    limit=H/2/np.cos(ang)
    z=np.r_[line(-limit,-1.001*a,N),line(1.001*a,limit,N)]*np.exp(1j*ang)
    f=evaluate_upper(case,z);fm=evaluate_upper(case,-z);diff=f-fm
    residuals["interface_displacement_rms"]=float(np.sqrt(np.mean((f[:,:2]+fm[:,:2])**2)))
    vals=np.c_[nx*diff[:,2]+ny*diff[:,4],nx*diff[:,4]+ny*diff[:,3]]
    residuals["interface_traction_rms"]=float(np.sqrt(np.mean(vals**2)))
    return residuals

def constitutive_check():
    mismatches=[]
    for mode in ("I","II"):
        for side in (1,-1):
            ang=np.deg2rad(30);orientation=ang+(np.pi if side==-1 else 0)
            z=side*a*np.exp(1j*ang)+np.exp(1j*orientation)*np.array([.3+side*.2j,.5+side*.4j])
            def f(zz):return will(zz,ang,side,mode,.5)
            delta=1e-6*a
            dx=(f(z+delta)-f(z-delta))/(2*delta)
            dy=(f(z+1j*delta)-f(z-1j*delta))/(2*delta)
            exx,eyy,gamma=dx[:,0],dy[:,1],dy[:,0]+dx[:,1]
            derived=np.c_[(lam+2*mu)*exx+lam*eyy,lam*exx+(lam+2*mu)*eyy,mu*gamma]
            exact=f(z)[:,2:]
            mismatches.append(np.linalg.norm(derived-exact)/np.linalg.norm(exact))
    return max(mismatches)
print("Unit enrichment maximum Hooke relative mismatch:",constitutive_check())
assert constitutive_check()<1e-7

for angle,case in cases.items():
    print(f"\nAngle {angle}: independent residuals:",independent_boundary_checks(case))
    r=a*np.geomspace(1e-6,1e-3,25);ang=case["angle_rad"];tip=a*np.exp(1j*ang)
    ahead=rotate(evaluate_upper(case,tip+r*np.exp(1j*ang)),-ang)
    face=rotate(evaluate_upper(case,tip-r*np.exp(1j*ang)),-ang)
    kis=np.sqrt(2*np.pi*r)*ahead[:,3]
    kiis=np.sqrt(2*np.pi*r)*ahead[:,4]
    t=np.sqrt(r/a);V=np.c_[np.ones_like(t),t,t**2,t**3,t**4]
    # Fit both local displacement components together; coefficient of sqrt(r/a).
    fit=lstsq(V,face[:,:2],cond=1e-13,lapack_driver="gelsd")[0]
    factor=2*mu/(kappa+1)*np.sqrt(2*np.pi/a)
    kid,kiid=factor*fit[1,1],factor*fit[1,0]
    case["checks"]=dict(r=r,KI_stress=kis,KII_stress=kiis,KI_displacement=kid,KII_displacement=kiid)
    print(f"  KI : coefficient={case['coeff'][-4]:.9f}, stress={kis[0]:.9f}, face-fit={kid:.9f}")
    print(f"  KII: coefficient={case['coeff'][-3]:.9f}, stress={kiis[0]:.9f}, face-fit={kiid:.9f}")
```

```python
# -*- coding: utf-8 -*-
"""Redraw saved mixed-mode source histories. No solver or training is run."""
from pathlib import Path
import csv
import numpy as np
import matplotlib.pyplot as plt
BASE_DIR = Path(__file__).resolve().parent if '__file__' in globals() else Path.cwd()
DATA_DIR = BASE_DIR / 'mixed_mode_adaptive_cold_results'
PLOT_ANGLE = 30
PLOT_CFG = dict(figsize=(20, 9.4), label_fontsize=17, tick_fontsize=13,
                legend_fontsize=15, line_width=1.8, save_dpi=400,
                save_outputs=True, show=True, show_best_markers=False)

def read_mixed_histories(folder, angle=30):
    folder=Path(folder)
    path=folder/'adaptive_source_initial_final_summary.csv'
    if not path.is_file():
        raise FileNotFoundError(f'Missing {path.resolve()}. Use the last notebook redraw cell while source_study is in memory to export histories first.')
    with path.open(newline='', encoding='utf-8-sig') as f:
        summaries=list(csv.DictReader(f))
    runs=[]
    for i,row in enumerate(summaries):
        if float(row['angle']) != float(angle): continue
        with (folder/f'adaptive_source_history_{angle:g}deg_run_{i+1}.csv').open(newline='',encoding='utf-8-sig') as f:
            history=list(csv.DictReader(f))
        runs.append(dict(adaptive_history=history, initial_offsets=(float(row['initial_offset_x']),float(row['initial_offset_y'])),
                         best_record=int(float(row.get('record',len(history)-1)))))
    with np.load(folder/f'adaptive_source_reference_{angle:g}deg.npz',allow_pickle=False) as d:
        refs=(float(d['KI']),float(d['KII'])); H=float(d['H'])
    return runs,refs,H

def plot_mixed_histories(runs, refs, H, output_dir, angle=30, settings=None):
    pc=dict(PLOT_CFG)
    if settings: pc.update(settings)
    if not runs: raise ValueError('No adaptive source histories available.')
    colors=['#1f77b4','#ff7f0e','#2ca02c','#d62728']
    panels=[('offset_x_over_H',r'$m_x/H$',False),('offset_y_over_H',r'$m_y/H$',False),
            ('KI',r'Stress intensity factor $K_I$',False),('KII',r'Stress intensity factor $K_{II}$',False),
            ('boundary_residual',r'Boundary residual $\sqrt{L_b}$',True),
            ('err_KI',r'Difference from BEM in $K_I$ (%)',True),
            ('err_KII',r'Difference from BEM in $K_{II}$ (%)',True),('total_loss','Total training loss',True)]
    style={'font.family':'serif','font.serif':['Times New Roman','DejaVu Serif'],
           'mathtext.fontset':'stix','font.size':13,'axes.linewidth':.85,'pdf.fonttype':42}
    with plt.rc_context(style):
        fig,axes=plt.subplots(2,4,figsize=pc['figsize'],dpi=150,constrained_layout=True)
        fig.set_constrained_layout_pads(w_pad=.07,h_pad=.07,wspace=.07,hspace=.07)
        for panel,(ax,(key,label,logscale)) in enumerate(zip(axes.ravel(),panels)):
            xmax=0
            for j,case in enumerate(runs):
                history=case['adaptive_history']
                x=np.asarray([float(r['record']) for r in history]);y=np.asarray([float(r[key]) for r in history])
                if not np.isfinite(x).all() or not np.isfinite(y).all(): raise ValueError(f'Non-finite values in {key}.')
                if key.startswith('err_'): y=y*100 # Stored errors are fractions.
                y=np.where(y>0,y,np.nan) if logscale else y
                mx,my=case['initial_offsets'];name=rf'$({mx/H:g},\,{my/H:g})$'
                ax.plot(x,y,color=colors[j%4],lw=pc['line_width'],label=name)
                xmax=max(xmax,float(x.max()))
                if pc['show_best_markers']:
                    k=next((k for k,r in enumerate(history) if int(float(r['record']))==int(case['best_record'])),None)
                    if k is not None: ax.plot(x[k],y[k],'o',ms=4.5,mfc='white',color=colors[j%4])
            if key in ('KI','KII'):
                ref=refs[0 if key=='KI' else 1]
                ax.axhline(ref,color='#444444',ls='--',lw=1.2,zorder=1)
                ax.annotate('BEM reference',xy=(.98,ref),xycoords=('axes fraction','data'),
                            xytext=(0,-8),textcoords='offset points',ha='right',va='top',fontsize=12,color='#444444')
            if logscale: ax.set_yscale('log')
            ax.set_xlim(0,xmax)
            ax.set_xlabel('Iteration',fontsize=pc['label_fontsize'])
            ax.set_ylabel(label,fontsize=pc['label_fontsize'])
            ax.tick_params(direction='in',top=True,right=True,labelsize=pc['tick_fontsize'])
            ax.grid(False)
            ax.margins(y=.12)
            if panel==0:
                ax.legend(loc='lower right',ncol=2,fontsize=pc['legend_fontsize'],frameon=False,
                          handlelength=1.5,columnspacing=.7,handletextpad=.4)
            elif panel!=1:
                ax.legend(loc='lower right' if key in ('KI','KII') else 'upper right',ncol=1,
                          fontsize=pc['legend_fontsize'],frameon=False,handlelength=1.5,handletextpad=.4)
        if pc['save_outputs']:
            folder=Path(output_dir);folder.mkdir(parents=True,exist_ok=True)
            for ext in ('png','pdf'):
                file=folder/f'adaptive_source_convergence_{angle:g}deg.{ext}'
                fig.savefig(file,dpi=pc['save_dpi'],bbox_inches='tight',facecolor='white')
                print('Saved:',file.resolve())
        if pc['show']: plt.show()
    return fig,axes


# Run only this cell if training results are already in memory.
if 'source_study' in globals() and source_study:
    folder=Path(cfg['output_dir']);folder.mkdir(parents=True,exist_ok=True)

    # Export all records for later redraws without retraining.
    for index,case in enumerate(source_study,1):
        history=case['adaptive_history']
        with (folder/f"adaptive_source_history_{cfg['plot_angle']:g}deg_run_{index}.csv").open('w',newline='',encoding='utf-8-sig') as f:
            writer=csv.DictWriter(f,fieldnames=list(history[0]));writer.writeheader();writer.writerows(history)
    np.savez(folder/f"adaptive_source_reference_{cfg['plot_angle']:g}deg.npz",
             KI=BEM_REF[cfg['plot_angle']][0]*sigma0,KII=BEM_REF[cfg['plot_angle']][1]*sigma0,H=H)
    plot_mixed_histories(source_study,tuple(v*sigma0 for v in BEM_REF[cfg['plot_angle']]),H,folder,cfg['plot_angle'])
else:
    runs,refs,plate_H=read_mixed_histories(DATA_DIR,PLOT_ANGLE)
    plot_mixed_histories(runs,refs,plate_H,DATA_DIR,PLOT_ANGLE)
```
